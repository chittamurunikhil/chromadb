```python
import chromadb

client = chromadb.Client()


```

```python
collection = client.get_or_create_collection (
    name= "my_search_collection_1"
)
```

```python
collection.add(
    ids=["id1","id2"],
    documents = ["this is first documnents", "this is another document"],
    embeddings = [[1.2,2.3],[3.4,4.5]]
)
```

```python

# query by vector similarity
results = collection.query(
    query_embeddings = [[3.1,4.2]],
    n_results = 1
)
```

```python
import chromadb
client = chromadb.Client()

collection = client.get_or_create_collection(
    name="my_collection"
)

# Add documents with embeddings
collection.add(
    ids=["id1", "id2"],
    documents=["This is a document", "Another doc"],
    embeddings=[[1.2, 2.3], [3.4, 4.5]]
)

# Query by vector similarity
results = collection.query(
    query_embeddings=[[1.1, 2.2]],
    n_results=10
)
```
